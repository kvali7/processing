```python
import numpy as np
import matplotlib.pyplot as plt
import sys
from pathlib import Path

project_root = (Path.cwd() / "..").resolve()
if str(project_root) not in sys.path:
    sys.path.append(str(project_root))
from slopes.misc import compute_epsilon

saturation = np.linspace(0.0, 1.0, num=100)
wv1_epsilon = compute_epsilon(saturation, wavelength=735.0)
wv2_epsilon = compute_epsilon(saturation, wavelength=850.0)
# epsilon_ratio = wv1_epsilon / wv2_epsilon
epsilon_ratio = wv1_epsilon / wv2_epsilon

# epsilon_ratio = compute_epsilon_ratio(saturation)

# plt.plot(saturation, wv1_epsilon, label="Epsilon (735 nm)")
# plt.plot(saturation, wv2_epsilon, label="Epsilon (850 nm)")
plt.plot(epsilon_ratio, saturation, label="Epsilon Ratio")
plt.legend()
plt.ylabel("Saturation")
plt.xlabel("Epsilon Ratio")
```
